```python
import pandas as pd

df=pd.read_csv("Global_Superstore_Cleaned.csv")
```

```python
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
df["Order Date"]= pd.to_datetime(df["Order Date"])
df["Ship Date"]= pd.to_datetime(df["Ship Date"])
df.head()
```

```python
print ("=" * 60)
print ("Global Superstore Data")
print ("=" * 60)

print (f"Total Sales     : ${df['Sales'].sum():,.2f}")
print (f"Total Profits   : ${df['Profit'].sum():,.2f}")
print (f"Total Orders    : {df['Order ID'].nunique():,}")
print (f"Total Customers : {df['Customer ID'].nunique():,}")
print (f"Total Products  : {df['Product Name'].nunique():,}")
print (f"Total Countrys  : {df['Country'].nunique():,}")
print (f"Total Markets   : {df['Market'].nunique():,}")
```

```python
df["Sales"].describe()
```

```python
# Top 10 Country's by Sales 
top_country = (
    df.groupby("Country")["Sales"]
        .sum()
        .sort_values(ascending=False)
        .head(10))
top_country
    
```

```python
# Visualization 

plt.figure(figsize=(12,6))
plt.bar(top_country.index,top_country.values)
plt.title("Top 10 Country's by Sales")
plt.xlabel("Country")
plt.ylabel("Sales")
plt.xticks(rotation =45)
plt.show()
```

```python
# Sales By Category 
top_sales = (
    df.groupby("Category")["Sales"]
        .sum()
        .sort_values(ascending=False))

#visualization
plt.figure(figsize=(8,4))
plt.bar(top_sales.index,top_sales.values)
plt.title("Sales By Category")
plt.ylabel("Sales")
plt.xlabel("Category")
plt.show()
top_sales
```

```python
# Profit By Category 
top_profits = (
    df.groupby("Category")["Profit"]
        .sum()
        .sort_values(ascending=False))

#visualization
plt.figure(figsize=(8,4))
plt.bar(top_profits.index,top_profits.values)
plt.title("Profit By Sales")
plt.ylabel("Profit")
plt.xlabel("Category")
plt.show()
top_profits
```

```python
# Top 10 Customers by Sales
top_customers = (
    df.groupby("Customer Name")["Sales"]
        .sum()
        .sort_values(ascending=False)
        .head(10))

#visualization
plt.figure(figsize=(17,6))
plt.bar(top_customers.index,top_customers.values)
plt.title("# Top 10 Customers by Sales")
plt.ylabel("Sales")
plt.xlabel("Customer Name")

plt.show()
top_customers
```

```python
#profits by Country
profits_by_country = (
    df.groupby("Country")["Profit"]
        .sum()
        .sort_values(ascending=False)
        .head(10))
#visualization
plt.figure(figsize=(17,10))
plt.bar(profits_by_country.index,profits_by_country.values)
plt.title("Top 10 most profitable countries")
plt.ylabel("Profit")
plt.xlabel("Country")

plt.show()
profits_by_country
```
